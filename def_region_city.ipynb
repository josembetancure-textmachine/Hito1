```python
#Intentaré cargar el DataFrame para extraer de aquí los municipios.

df_region_city=pd.read_csv("./datos_sucios_hito1.csv")
#df_region_city

#1
#Definiré un nuevo DataFrame, a partir del anterior, para visualizar únicamente los municipios
#city_columns=["Municipio"]
# df_city=pd.read_csv("./datos_sucios_hito1.csv",usecols=city_columns)
# df_city

# 2. Opción ganadora por eficiencia de recursos
# df_city=df_region_city[["Municipio"]]
# df_city

#3. Sin embargo, puedo definir la lista de municipios directamente
city_list=df_region_city["Municipio"].unique().tolist()
city_list

#4. Ahora que tengo la lista con los municipios, puedo generar una función para crear el diccionario
def dic(city_list,progress=None): #Estoy tratando de crear un diccionario con city como llave y región como valor.
    
    if progress is None:
        city_region_dic={}
    else:
        city_region_dic=progress.copy()

    for c in city_list:

        if c in city_region_dic:
            continue

        while True:
            region=input(f"Ingrese una región para {c}: \n O ingrese '-' para terminar.").strip().lower()
            
            if region=="-":
                return city_region_dic
            elif region=="":
                print(f"Debe ingresar una región para {c}")
                continue #Es decir, que se repita la pregunta para cada municipio si no se ingresó nada.
            else:
                break

        if region=="uraba": #Solo con este porque es la única región que lleva tilde
            region=region.replace("uraba","urabá")

        region=region.capitalize()
        print(f"Municipio: {c}, región: {region}")
        city_region_dic[c]=region

    return city_region_dic
cpoint=dic(city_list)
rpoint=dic(city_list,progress=cpoint)

print(rpoint)
```
